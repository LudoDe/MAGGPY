```python
import emcee
import matplotlib.pyplot        as plt
import numpy                    as np
from scipy.optimize             import curve_fit
from pathlib                    import Path
from src.top_hat.geometric_eff  import geometric_efficiency_fixed, create_geometric_efficiency_lognormal_interpolator, geometric_efficiency_flat_midpoint
from src.redshift               import get_mrd_redshift_distribution
from scipy.stats                import binned_statistic
from scipy.optimize             import brentq
from mcmc_runner                import _backend_path

def calculate_theta_star_from_beaming(theta_samples, fj_samples, beaming_func, boundary_limit=9.0, n_bins=50):
    """
    Calculates theta* (where f_j = 1).
    """
    min_theta, max_theta = np.min(theta_samples), np.max(theta_samples)        
    bins = np.logspace(np.log10(min_theta), np.log10(max_theta), n_bins + 1)
    
    # Calculate 95th Percentile in each bin
    p95_per_bin, bin_edges, bin_indices = binned_statistic(
        theta_samples, fj_samples, 
        statistic=lambda y: np.percentile(y, 95), 
        bins=bins
    )
    
    valid_bin_indices = np.where(p95_per_bin < boundary_limit)[0] + 1
    
    if len(valid_bin_indices) == 0:
        print("No valid samples found within boundary limit.")
        return np.nan, np.nan, np.zeros(len(theta_samples), dtype=bool)

    # Calculate Constant
    valid_samples_mask = np.isin(bin_indices, valid_bin_indices)

    theta_valid = theta_samples[valid_samples_mask]
    fj_valid    = fj_samples[valid_samples_mask]

    beaming_vals = beaming_func(theta_valid)
    C_est        = np.median(fj_valid * beaming_vals)
    
    def objective(t):
        return beaming_func(t) - C_est
        
    # Determine search range respecting function bounds
    lower, upper = 1.1, 45
    if hasattr(beaming_func, 'x'): 
        lower = max(lower, beaming_func.x[0])
        upper = min(upper, beaming_func.x[-1])

    theta_star = np.nan
    val_lower = objective(lower)
    val_upper = objective(upper)
    
    if np.isnan(val_lower) or np.isnan(val_upper):
        return np.nan, C_est, valid_samples_mask
    
    if val_lower * val_upper < 0:
        theta_star = brentq(objective, lower, upper)

    return theta_star, C_est, valid_samples_mask

#cache the results of the function to avoid redundant calculations
def get_single_theta_star(backend, geom_eff_func, fj_cutoff=9.0):
    samples             = backend.get_chain(discard=5000, flat=True)
    theta_c_samples     = samples[:, 4]
    fj_samples          = samples[:, 5]
    theta_star, _, _    = calculate_theta_star_from_beaming(theta_c_samples, fj_samples, geom_eff_func, boundary_limit=fj_cutoff)
    return theta_star

#alphas = ["A0.5", "A1.0", "A3.0", "A5.0"]

#datafiles = Path("../../datafiles")
#alpha_to_rate = {}
##alpha_to_total_rate = {}
#for alpha in alphas:
#    _, total_rate, local_rate, _, _ = get_mrd_redshift_distribution(datafiles, "fiducial_delayed", alpha, "BNSs")
#    alpha_to_rate[alpha] = local_rate
#    alpha_to_total_rate[alpha] = total_rate

#lognorm_geom = create_geometric_efficiency_lognormal_interpolator(0.5, 200, 1.0, 25.0)

#geometries = [
#    ("universal", geometric_efficiency_fixed, "o"),
#    ("lognormal", lognorm_geom, "s"),
#    ("flat_midpoint", geometric_efficiency_flat_midpoint, "^")
#]

#dictionary_thetas = {}

#for run_dir, geom_func, marker in geometries:
#    alpha_val, theta_cross  = analyse_crossover_alpha(alphas, geom_func, run_dir)
#    dictionary_thetas[run_dir] = theta_cross
```

```python
# Diagnostic plots: fraction > 10 deg and NSBH relative contribution
theta_threshold     = 10.0
alpha_tests         = ["A0.5", "A1.0", "A3.0", "A5.0"]

theta_bins = np.linspace(1.0, 25.0, 10) # for the theta_nsbh binning
bin_size = theta_bins[1] - theta_bins[0]
def get_theta_stars(alpha, geom_eff_func):
    backend_1       = emcee.backends.HDFBackend(_backend_path(alpha, stage=1))
    backend_2       = emcee.backends.HDFBackend(_backend_path(alpha, stage=2))
    samples_bns_only = backend_1.get_chain(discard=5000, flat=True)
    samples_bns_nsbh = backend_2.get_chain(discard=5000, flat=True)
    
    fj_samples_bns_only             = samples_bns_only[:, -1]
    theta_bns_samples_bns_only      = samples_bns_only[:, -2] #last two

    fj_samples_bns_nsbh             = samples_bns_nsbh[:, -1]
    theta_bns_samples_bns_nsbh      = samples_bns_nsbh[:, 0]
    theta_nsbh_samples_bns_nsbh     = samples_bns_nsbh[:, 1]
    
    theta_star_bns_only, _, _ = calculate_theta_star_from_beaming(theta_bns_samples_bns_only, fj_samples_bns_only, geom_eff_func)

    theta_star_bns_nsbh = []
    for theta_nsbh_i in theta_bins:
        mask_low = theta_nsbh_samples_bns_nsbh > theta_nsbh_i - bin_size/2
        mask_high = theta_nsbh_samples_bns_nsbh < theta_nsbh_i + bin_size/2
        mask = mask_low & mask_high 
        fj_samples_nsbh_i           = fj_samples_bns_nsbh[mask]
        theta_bns_samples_nsbh_i    = theta_bns_samples_bns_nsbh[mask]
        theta_star_nsbh_i, _, _ = calculate_theta_star_from_beaming(theta_bns_samples_nsbh_i, fj_samples_nsbh_i, geom_eff_func)
        theta_star_bns_nsbh.append(theta_star_nsbh_i)

    return theta_star_bns_only, theta_star_bns_nsbh


for alpha_val in alpha_tests:
    theta_s_bns_only, theta_s_bns_nsbh = get_theta_stars(alpha_val, geom_eff_interp)
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(theta_bins, theta_s_bns_nsbh, label='BNS+NSBH', marker='o')
    ax.axhline(theta_s_bns_only, color='r', linestyle='--', label='BNS-only')
    ax.text(0.5, 0.9, f'{alpha_val}', transform=ax.transAxes, fontsize=12, ha='center')
    ax.set_xlabel(r'$\theta_{\mathrm{NSBH}}$ (degrees)')
    ax.set_ylabel(r'$\theta^*$ (degrees)')
    ax.legend()
    plt.show()
    
```
